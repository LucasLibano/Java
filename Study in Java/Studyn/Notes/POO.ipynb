- A POO (Programação Orientada a Objetos) em Java,  é um paradigma de programação baseado em objetos, que representam coisas do mundo real dentro do código.

- Programação Orientada a Objetos é uma forma de programar onde você organiza o código usando:

Classes = Pessoa

Objetos = João

Atributos = Atributos (Caracteristcas).

Métodos = Metodos (Ações).

- Classe

Uma classe é um molde (modelo) para criar objetos.

Exemplo:

public class Produto {
    public String nome;
    public double preco;
    public int quantidade;
}

Aqui criamos a classe Produto.

- Objeto

O objeto é uma instância da classe.

Exemplo:

Produto p1 = new Produto();

Agora p1 é um objeto da classe Produto.

Podemos atribuir valores:

p1.nome = "Notebook";
p1.preco = 3000.00;
p1.quantidade = 5;

- Atributos

São as variáveis da classe.

No exemplo:

public String nome;
public double preco;
public int quantidade;

Esses atributos guardam informações do objeto.

- Métodos

São funções dentro da classe que definem comportamentos.

Exemplo:

public double valorTotal() {
    return preco * quantidade;
}

Uso:

double total = p1.valorTotal();

Exemplo completo: 

Classe:

public class Produto {
    public String nome;
    public double preco;
    public int quantidade;

    public double valorTotalEmEstoque() {
        return preco * quantidade;
    }
}

Main:

Produto produto = new Produto();

produto.nome = "Notebook";
produto.preco = 3000.0;
produto.quantidade = 10;

System.out.println(produto.valorTotalEmEstoque());

Os 4 pilares da POO (que Nelio ensina depois)

1️⃣ Encapsulamento
Controlar o acesso aos dados.

2️⃣ Herança
Uma classe herdar características de outra.

3️⃣ Polimorfismo
Um método ter vários comportamentos.

4️⃣ Abstração
Representar apenas o que é importante.
