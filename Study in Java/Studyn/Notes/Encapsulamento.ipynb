- Encapsulamento.

É um principio que consiste em esconder detalhes de implementação de uma classe, expondo apenas operações seguras e que mantenhham os objetos em um estados consistente.

O objeto deve sempre estar em um estado consistente, e a própria classe deve garantir isso.

O objeto não deve expor nenum atributo (modificador de acesso private). Os atributos devem ser acessados por meios de metodos GET e SET.

- Modificadores de acesso.

Private: O membro só pode ser acessado nas classes do mesmo pacote.

(nada): O membro só pode ser acessado nas classes do mesmo pacote.

protected: o membro só pode ser acessado no mesmo pacote, bem como em subclasses de pacotes diferentes.

public: o membro é acessado por todas classes (ao menos que ele resida em um módulo diferente que não exporte o pacote onde ele está).
